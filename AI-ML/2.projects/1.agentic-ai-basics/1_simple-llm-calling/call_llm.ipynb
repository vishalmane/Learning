```python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()
```

```python
llm = ChatGoogleGenerativeAI(model="gemini-3.5-flash")

response=llm.invoke("How many moons does jupiter hava?")
print(response.text)
```

```python
response=llm.invoke([
    ["system","you are a healpful assistant that answers in one line"],
    ["human","How many moons does jupiter hava?"]
])
print(response.text)

```

```python
llm = ChatGoogleGenerativeAI(model="gemini-3.5-flash", temperature=0)
response=llm.invoke("Tell me architect joke")
print(response.text)

```
